```python
"""
Python DA Assignment 1: Data Analysis using NumPy and Pandas
================================================================
Complete solution with line-by-line explanations.
"""

# ============================================================
# SECTION 1: NUMPY ARRAY OPERATIONS
# ============================================================

import numpy as np   # Import the NumPy library and alias it as 'np' (standard convention)

# ------------------------------------------------------------
# Step 1: Create a 1D NumPy Array
# ------------------------------------------------------------

# np.array() converts a Python list into a NumPy array object,
# which enables fast, vectorized numerical operations.
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:", temperatures_w1)


# ------------------------------------------------------------
# Step 2: Inspection and Properties
# ------------------------------------------------------------

# .shape returns a tuple showing the array's dimensions (rows, columns).
# For a 1D array of 7 elements, this will be (7,)
print("\nShape of temperatures_w1:", temperatures_w1.shape)

# .dtype returns the data type of the elements stored in the array (e.g., float64).
print("Data type of temperatures_w1:", temperatures_w1.dtype)

# .size returns the total number of elements in the array.
print("Number of elements in temperatures_w1:", temperatures_w1.size)


# ------------------------------------------------------------
# Step 3: Array Operations
# ------------------------------------------------------------

# NumPy allows vectorized arithmetic: the formula is applied to every
# element simultaneously without needing an explicit loop.
temperatures_w1_fahrenheit = (temperatures_w1 * 9 / 5) + 32
print("\nWeek 1 Temperatures in Fahrenheit:", temperatures_w1_fahrenheit)

# .max(), .min(), and .mean() are built-in NumPy aggregate functions
# that compute the maximum, minimum, and average of all elements.
max_temp = temperatures_w1.max()
min_temp = temperatures_w1.min()
mean_temp = temperatures_w1.mean()

print("Maximum Temperature (Week 1):", max_temp)
print("Minimum Temperature (Week 1):", min_temp)
print("Mean Temperature (Week 1):", round(mean_temp, 2))


# ------------------------------------------------------------
# Step 4: Array Slicing and Indexing
# ------------------------------------------------------------

# Slicing syntax array[start:stop] extracts elements from index 'start'
# up to (but not including) index 'stop'.

# First three days -> indices 0, 1, 2
first_three_days = temperatures_w1[0:3]
print("\nFirst three days:", first_three_days)

# Weekend -> last two days -> negative indexing counts from the end
# -2 means "2nd from last" and omitting 'stop' means "till the end"
weekend_temps = temperatures_w1[-2:]
print("Weekend temperatures:", weekend_temps)

# Middle three days -> indices 2, 3, 4 (days 3, 4, and 5 of the week)
middle_three_days = temperatures_w1[2:5]
print("Middle three days:", middle_three_days)


# ------------------------------------------------------------
# Step 5: Create a 2D Array
# ------------------------------------------------------------

# A 2D array is created by passing a list of lists to np.array().
# Each inner list becomes a row in the resulting matrix.
temperatures_w2 = np.array([19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1])

temperatures = np.array([
    temperatures_w1,   # Row 1: Week 1 data
    temperatures_w2    # Row 2: Week 2 data
])

print("\n2D Temperatures Array (Week 1 & Week 2):\n", temperatures)


# ------------------------------------------------------------
# Step 6: Inspect and Slice the 2D Array
# ------------------------------------------------------------

# For a 2D array, .shape returns (rows, columns) -> here (2, 7)
print("\nShape of temperatures (2D):", temperatures.shape)
print("Data type of temperatures (2D):", temperatures.dtype)
print("Total number of elements:", temperatures.size)

# temperatures[0] selects the entire first row -> Week 1 data
week1_data = temperatures[0]
# temperatures[1] selects the entire second row -> Week 2 data
week2_data = temperatures[1]

print("\nWeek 1 data:", week1_data)
print("Week 2 data:", week2_data)

# temperatures[0, -2:] -> row 0, last two columns -> Week 1 weekend
week1_weekend = temperatures[0, -2:]
# temperatures[1, -2:] -> row 1, last two columns -> Week 2 weekend
week2_weekend = temperatures[1, -2:]

print("Week 1 Weekend temperatures:", week1_weekend)
print("Week 2 Weekend temperatures:", week2_weekend)
```

```python
# ============================================================
# SECTION 2: PANDAS SERIES
# ============================================================

import pandas as pd   # Import the Pandas library and alias it as 'pd' (standard convention)

# ------------------------------------------------------------
# Step 1: Creating a Pandas Series
# ------------------------------------------------------------

# pd.Series() creates a one-dimensional labeled array.
# 'data' holds the values, 'index' holds the custom row labels.
marks = pd.Series(
    data=[95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("\n\nMarks Series:\n", marks)


# ------------------------------------------------------------
# Step 2: Indexing and Slicing
# ------------------------------------------------------------

# Using the integer position (0-based) to access the 1st element,
# regardless of the custom label assigned to it.
# NOTE: In modern pandas, plain positional indexing with marks[0] is
# deprecated/removed for label-indexed Series, so .iloc[] is the
# correct, future-proof way to do this.
rank1_mark_by_position = marks.iloc[0]
print("\nMark of Rank1 (via integer position):", rank1_mark_by_position)

# .loc[] accesses data using LABELS. Passing a list of labels
# returns a Series containing only those rows, in that order.
top3_marks = marks.loc[['Rank1', 'Rank2', 'Rank3']]
print("Top 3 ranks' marks (loc):\n", top3_marks)

# .iloc[] accesses data using integer POSITIONS (like a normal list).
# Position 2 corresponds to the 3rd element (0-indexed).
rank3_mark = marks.iloc[2]
print("Mark of Rank3 student (iloc):", rank3_mark)

# Boolean masking: marks > 90 creates a Series of True/False values;
# passing this mask back into marks[] filters rows where it's True.
high_scorers = marks[marks > 90]
print("Ranks with marks greater than 90:\n", high_scorers)


# ------------------------------------------------------------
# Step 3: Manipulating the Series
# ------------------------------------------------------------

# Direct label-based assignment updates the value at 'Rank1' to 100.
marks['Rank1'] = 100
print("\nMarks after updating Rank1 to 100:\n", marks)

# .drop() removes the specified label from the Series.
# By default, this returns a NEW Series (doesn't modify in place)
# unless inplace=True is passed.
marks = marks.drop('Rank5')
print("Marks after removing Rank5:\n", marks)

# Dividing the whole Series by 10 is a vectorized operation -
# it applies to every element at once, producing the CGPA scale.
cgpa = marks / 10
print("CGPA (Marks / 10):\n", cgpa)
```

```python
# ============================================================
# SECTION 3: PANDAS DATAFRAME
# ============================================================

# ------------------------------------------------------------
# Step 1: Creating a Pandas DataFrame
# ------------------------------------------------------------

# A DataFrame can be built from a dictionary where each key becomes
# a column name and each value (a list) becomes that column's data.
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                         'Clothing', 'Electronics', 'Furniture', 'Clothing',
                         'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print("\n\nTransactions DataFrame:\n", transactions)


# ------------------------------------------------------------
# Step 2: Data Exploration
# ------------------------------------------------------------

# .head() shows the first 5 rows by default - useful for a quick preview.
print("\nHead (first 5 rows):\n", transactions.head())

# .tail() shows the last 5 rows by default.
print("\nTail (last 5 rows):\n", transactions.tail())

# .shape returns (number_of_rows, number_of_columns).
print("\nShape of DataFrame:", transactions.shape)

# .columns lists all column names as an Index object.
print("Column names:", transactions.columns.tolist())

# .dtypes shows the data type stored in each column.
print("Data types:\n", transactions.dtypes)

# .info() prints a concise summary: dtypes, non-null counts, memory usage.
print("\nDataFrame Info:")
transactions.info()

# Selecting multiple columns: pass a LIST of column names inside [].
category_amount = transactions[['ProductCategory', 'Amount']]
print("\nProductCategory and Amount columns:\n", category_amount)

# .iloc[:, -3:] -> all rows (:), last 3 columns (-3:)
last_3_cols = transactions.iloc[:, -3:]
print("\nLast 3 columns (iloc):\n", last_3_cols)

# Boolean filtering with multiple conditions requires '&' (and) / '|' (or),
# and each condition must be wrapped in parentheses due to operator precedence.
north_above_200 = transactions[(transactions['Region'] == 'North') &
                                (transactions['Amount'] > 200)]
print("\nNorth region transactions with Amount > 200:\n", north_above_200)

# .value_counts() counts the frequency of each unique value in a column.
category_counts = transactions['ProductCategory'].value_counts()
print("\nValue counts for ProductCategory:\n", category_counts)

# .unique() returns an array of distinct values in a column.
unique_regions = transactions['Region'].unique()
print("\nUnique values in Region:", unique_regions)

# .groupby('Region') groups rows sharing the same Region value;
# ['Amount'].mean() then computes the average Amount within each group.
avg_amount_by_region = transactions.groupby('Region')['Amount'].mean()
print("\nMean Amount by Region:\n", avg_amount_by_region)


# ------------------------------------------------------------
# Step 3: Manipulating the DataFrame
# ------------------------------------------------------------

# .loc[row_condition, 'column'] locates the row(s) matching the condition
# and sets the value in the specified column for those rows.
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print("\nAfter updating Amount for TransactionID 102:\n", transactions)

# Assigning to a new column name creates it; here each value is
# computed as 10% of the corresponding 'Amount' value (vectorized).
transactions['Discount'] = transactions['Amount'] * 0.10
print("\nAfter adding Discount column:\n", transactions)

# To remove a row, first find its index label using boolean filtering,
# then pass that index to .drop(). inplace=True modifies the DataFrame directly.
transactions.drop(transactions[transactions['TransactionID'] == 109].index,
                   inplace=True)
print("\nAfter removing TransactionID 109:\n", transactions)

# .drop(columns=[...]) removes one or more columns.
# axis=1 (or columns=) specifies we're dropping a column, not a row.
transactions.drop(columns=['Discount'], inplace=True)
print("\nAfter deleting Discount column:\n", transactions)
```
